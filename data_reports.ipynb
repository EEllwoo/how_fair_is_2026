### What This section does:
Whilst checking which papers are present / missing from each pass I encountered as issue with paper identification.
Names have been copied over differently (a number have em dashes in, think people were copying the name from chatgpt) or have leading spaces and such. Paper dois also differe as some have the sheffield header and some are from non IEEE locations such as arXiv which have different dois. I have scraped a correct list of paper names and IEEE dois which passes are compared to and dois are then re-assiged properly such that papers can be easily identified between passes. To re-run the scraper you will need to upload your own IEEE sheffield cookies to run, but it has already been run and the list is correct so shouldnt be any need to. A few papers in Emmas first pass are missing dois so if dois are not availible it identifies papers by their name with hyhpens (,em dashes) and trailing/leading white space removed. As far as I am aware this correctly identifies all papers in the first and all papers missing in the second pass.  

```python
results_version = ["First","Second"]  # "first" or "second"
from pre_reports import find_missing_dois, missing_papers


for title in results_version:
    results_title = f"FAIR Evaluation Results - {title} Pass Responses"
    print(f"\n{'-'*8}{title} pass pre- report{'-'*8}")
    fixed = missing_papers.find_and_fix_missing_papers(
        tsv_path="results/paper_titles.tsv",
        csv_path=f"results/{results_title}.csv",
        output_csv_path=f"results/{results_title}_fixed.csv"
    )
    print(f"Corrected {len(fixed)} rows: {fixed}")
    
    find_missing_dois.doi_report(result_csv=f"results/{results_title}_fixed.csv")
```
