# Pass Comparison

Compare how the first and second review passes marked the same paper.

This notebook supports two views:
- all shared fields across both pass files
- FAIR-only fields using the F, A, I, and R criteria

```python
import importlib

from pre_reports import pass_compare_backend

importlib.reload(pass_compare_backend)

DEFAULT_FIRST_PASS = pass_compare_backend.DEFAULT_FIRST_PASS
DEFAULT_SECOND_PASS = pass_compare_backend.DEFAULT_SECOND_PASS
compare_all_papers = pass_compare_backend.compare_all_papers
compare_paper_marks = pass_compare_backend.compare_paper_marks
export_conflict_report = pass_compare_backend.export_conflict_report
get_overlap_summary = pass_compare_backend.get_overlap_summary
summarize_disagreements = pass_compare_backend.summarize_disagreements

first_pass_path = DEFAULT_FIRST_PASS
second_pass_path = DEFAULT_SECOND_PASS

overlap_summary = get_overlap_summary(first_pass_path, second_pass_path)
overlap_summary
```

```python
paper_name = "Fairquant: Certifying and Quantifying Fairness of Deep Neural Networks"

all_fields_for_paper = compare_paper_marks(
    paper_name,
    first_pass_path,
    second_pass_path,
    scope="all",
    differences_only=True,
)

fair_fields_for_paper = compare_paper_marks(
    paper_name,
    first_pass_path,
    second_pass_path,
    scope="fair",
    differences_only=True,
)

all_fields_for_paper, fair_fields_for_paper
```

```python
all_field_differences = compare_all_papers(
    first_pass_path,
    second_pass_path,
    scope="all",
    differences_only=True,
)

fair_field_differences = compare_all_papers(
    first_pass_path,
    second_pass_path,
    scope="fair",
    differences_only=True,
)

all_field_differences.head(20), fair_field_differences.head(20)
```

```python
conflict_report = export_conflict_report(
    first_pass_path=first_pass_path,
    second_pass_path=second_pass_path,
    top_n=10,
)

paper_conflict_report = conflict_report["paper_conflict_report"]
fair_conflict_report = conflict_report["fair_conflict_report"]
compact_columns = ["Paper Name:", "Differences", "FAIR Differences", "FAIR Letters"]
report_preview = {
    "overview": conflict_report["overview"],
    "markdown_report": str(conflict_report["output_paths"]["markdown_report"]),
}

(
    report_preview,
    paper_conflict_report[compact_columns].head(10),
    fair_conflict_report[compact_columns].head(10),
)
```
